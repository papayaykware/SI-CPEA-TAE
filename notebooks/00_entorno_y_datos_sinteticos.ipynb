## Entorno y datos sintéticos reproducibles

Este notebook comprueba el entorno y genera datos sintéticos ilustrativos. No representa datos humanos ni evidencia sobre sistemas superinteligentes.

```python
import platform
import numpy as np
import pandas as pd
SEED = 20261125
rng = np.random.default_rng(SEED)
print('Python:', platform.python_version())
print('NumPy:', np.__version__)
print('Pandas:', pd.__version__)
print('Seed:', SEED)
```

## Generación de variables sintéticas
Las distribuciones son arbitrarias y solo sirven para probar el flujo de trabajo.

```python
n = 500
df = pd.DataFrame({
    'trial': np.arange(n),
    'human_task_score': rng.normal(0.60, 0.10, n).clip(0, 1),
    'artificial_task_score': rng.normal(0.85, 0.07, n).clip(0, 1),
    'constraint_compliance': rng.beta(8, 2, n),
    'recursive_update': rng.integers(0, 6, n),
    'perturbation': rng.choice([0, 1], n, p=[0.7, 0.3])
})
df['capability_asymmetry'] = df['artificial_task_score'] - df['human_task_score']
df.head()
```

```python
assert len(df) == n
assert df.isna().sum().sum() == 0
assert df['capability_asymmetry'].between(-1, 1).all()
df.describe().round(3)
```
